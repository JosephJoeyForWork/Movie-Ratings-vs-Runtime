```python
import pandas as pd
from pathlib import Path

#Main Folder
folder = Path(r"C:\Users\Joseph\Desktop\LinkedIn Projects\Movie Ratings\movie data")

#find every .tsv file
for tsv_file in folder.rglob("*.tsv"):
    print(f"Converting {tsv_file}")

    df = pd.read_csv(tsv_file, sep="\t")

    #save .csv to the same file as .tsv
    csv_file = tsv_file.with_suffix(".csv")
    df.to_csv(csv_file, index=False)

print("All files converted successfully!")
```
